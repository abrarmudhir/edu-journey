### What is a turning point?

A **turning point** is the point on a graph where the curve **changes direction**.

Imagine you are riding a bike down a hill:

- You travel **downhill**
- You reach the **bottom**
- Then you start travelling **uphill**

The bottom is the **turning point** because this is where the direction changes.

For the graph:

$$
y = x^2 + x - 4
$$

the graph is shaped like a **U**.

So the turning point is the **lowest point of the U**.

For this graph, the turning point is:

$$
\boxed{(-0.5,\,-4.25)}
$$

This means:

- move \(0.5\) to the **left** on the \(x\)-axis
- move \(4.25\) **down** on the \(y\)-axis

and you reach the lowest point of the graph.

### Easy way to remember

**Turning point = where the graph stops going one way and starts going the other way.**

### 4(a) Chocolates and ratios

There are 280 chocolates in a box. There are only dark chocolates, milk chocolates and white chocolates.

$\frac{1}{7}$ of the 280 chocolates are dark chocolates.

The number of milk chocolates : the number of white chocolates $= 1:3$.

The number of white chocolates : the number of dark chocolates $= n:1$.

Work out the value of $n$. Show all your working.

### Worked solution

Dark chocolates: $\frac{1}{7}\times280=40$.

Milk and white chocolates together: $280-40=240$.

The milk:white ratio $1:3$ has $1+3=4$ equal parts. One part is $240\div4=60$.

Milk chocolates: $1\times60=60$. White chocolates: $3\times60=180$.

White:dark is $180:40=\frac{180}{40}:1=4.5:1$.

Therefore **$\boxed{n=4.5}$** (equivalently, $\frac{9}{2}$).

Check: $40+60+180=280$.

```python
from fractions import Fraction

total = 280
dark = total // 7
remaining = total - dark

milk_parts, white_parts = 1, 3
one_part = remaining // (milk_parts + white_parts)
milk = milk_parts * one_part
white = white_parts * one_part

n = Fraction(white, dark)
print(f"Dark = {dark}, milk = {milk}, white = {white}")
print(f"n = {n} = {float(n)}")

assert dark + milk + white == total
```

### Python code walkthrough

1. The total is stored as 280. Dividing it by 7 gives 40 dark chocolates. The double slash means whole-number division.
2. Subtracting the dark chocolates leaves 240 milk and white chocolates.
3. The milk:white ratio has $1+3=4$ parts. Dividing 240 by 4 gives 60 chocolates per part.
4. Multiplying by 1 and 3 gives 60 milk and 180 white chocolates.
5. Fraction(white, dark) calculates $180/40$ exactly as $\frac{9}{2}$; float(n) displays 4.5.
6. The final assertion checks that the counts add to 280.

## Question 5

Work out $5.7\times10^2+9.8\times10^3$.

Give your answer in **standard form**. (3 marks)

### Worked solution

First write both terms as ordinary numbers:

$$5.7\times10^2=5.7\times100=570$$

$$9.8\times10^3=9.8\times1000=9800$$

Add them:

$$570+9800=10370$$

In standard form, write the number as $a\times10^n$, where $1\leq a<10$. Move the decimal point four places to the left:

$$10370=\boxed{1.037\times10^4}$$

Check: $1.037\times10000=10370$.

```python
from decimal import Decimal

first = Decimal("5.7") * 10**2
second = Decimal("9.8") * 10**3
total = first + second

exponent = total.adjusted()
coefficient = total.scaleb(-exponent).normalize()

print(f"{first} + {second} = {total}")
print(f"Standard form: {coefficient} × 10^{exponent}")

assert total == 10370
assert coefficient == Decimal("1.037")
assert exponent == 4
```

### Python code walkthrough

1. Decimal stores 5.7 and 9.8 exactly. The powers 10**2 and 10**3 mean $10^2=100$ and $10^3=1000$.
2. The first two calculations give 570 and 9800. Adding them gives 10370.
3. adjusted() finds the exponent of the largest place value in 10370, which is 4.
4. scaleb(-exponent) moves the decimal point four places left to give the coefficient 1.037.
5. The printed result is $1.037\times10^4$. The assertions check the ordinary number, coefficient and exponent.

## Question 6

$AB$, $BC$, $CD$ and $DE$ are four consecutive sides of a regular polygon with $n$ sides.

$BCX$ is an equilateral triangle. $CDYX$ is a square.

Work out the value of $n$. Show all your working.

The diagram below is a schematic of the given construction.

```python
import math
import matplotlib.pyplot as plt
from matplotlib.patches import Arc

# Use a side length of 1. The exact size does not affect the angles.
C = (0.0, 0.0)
D = (0.0, -1.0)
X = (1.0, 0.0)
Y = (1.0, -1.0)
B = (0.5, math.sqrt(3) / 2)
A = (B[0] + math.cos(math.radians(30)),
     B[1] + math.sin(math.radians(30)))
E = (D[0] + math.cos(math.radians(300)),
     D[1] + math.sin(math.radians(300)))

fig, ax = plt.subplots(figsize=(6, 7))
def segment(p, q, color="black", width=2.2):
    ax.plot([p[0], q[0]], [p[1], q[1]],
            color=color, linewidth=width, solid_capstyle="round")

# Four consecutive sides of the regular polygon.
for p, q in [(A, B), (B, C), (C, D), (D, E)]:
    segment(p, q)

# Remaining triangle and square edges.
for p, q in [(B, X), (C, X), (X, Y), (Y, D)]:
    segment(p, q, color="#3b70a5", width=1.8)

offsets = {"A": (0.06, 0.02), "B": (-0.08, 0.08),
           "C": (-0.14, 0.02), "D": (-0.14, -0.07),
           "E": (0.02, -0.10), "X": (0.04, 0.02),
           "Y": (0.04, -0.07)}
for name, p in {"A": A, "B": B, "C": C, "D": D,
                "E": E, "X": X, "Y": Y}.items():
    dx, dy = offsets[name]
    ax.text(p[0] + dx, p[1] + dy, name, fontsize=15)

# At C, the 60-degree triangle angle and 90-degree square angle
# together form the polygon's 150-degree interior angle.
ax.add_patch(Arc(C, 0.38, 0.38, theta1=0, theta2=60,
                 color="#b54b3a", linewidth=1.8))
ax.add_patch(Arc(C, 0.58, 0.58, theta1=270, theta2=360,
                 color="#31805a", linewidth=1.8))
ax.text(0.21, 0.12, "60°", color="#b54b3a", fontsize=11)
ax.text(0.19, -0.23, "90°", color="#31805a", fontsize=11)

ax.set_aspect("equal")
ax.set_xlim(-0.45, 1.65)
ax.set_ylim(-2.1, 1.6)
ax.axis("off")
ax.set_title("Question 6: angle at C", fontsize=14)
plt.show()
```

### Worked solution

Because $BCX$ is an **equilateral triangle**, each of its angles is $60^\circ$. Therefore

$$\angle BCX=60^\circ.$$

Because $CDYX$ is a **square**, its angle at $C$ is

$$\angle XCD=90^\circ.$$

The interior angle of the regular polygon at $C$ is the angle between $CB$ and $CD$. The triangle and square split it into these two parts:

$$\angle BCD=60^\circ+90^\circ=150^\circ.$$

An exterior angle is supplementary to the interior angle:

$$180^\circ-150^\circ=30^\circ.$$

The exterior angles of any polygon add to $360^\circ$. All exterior angles of a regular polygon are equal, so

$$n=\frac{360^\circ}{30^\circ}=\boxed{12}.$$

The regular polygon has **12 sides**.

```python
triangle_angle = 60
square_angle = 90
interior_angle = triangle_angle + square_angle
exterior_angle = 180 - interior_angle
n = 360 // exterior_angle

print(f"Interior angle = {interior_angle}°")
print(f"Exterior angle = {exterior_angle}°")
print(f"Number of sides n = {n}")

assert interior_angle == 150
assert exterior_angle == 30
assert n == 12
```

### Python code walkthrough

1. The equilateral triangle gives a $60^\circ$ angle at $C$, and the square gives a $90^\circ$ angle at $C$.
2. Adding them gives the polygon's interior angle: $150^\circ$.
3. Subtracting from $180^\circ$ gives the exterior angle: $30^\circ$.
4. A complete turn is $360^\circ$. Dividing by $30^\circ$ gives 12 equal exterior angles, so the polygon has 12 sides.
5. The assertions check each result. The diagram code draws the four polygon sides, triangle and square, and labels the two angles at $C$.

### 7(a) Simplify

There is one factor of $(2-m)$ in the numerator and one in the denominator. Cancel the common factor:

$$\frac{3(2-m)^2}{2-m}
=\frac{3(2-m)(2-m)}{2-m}
=\boxed{3(2-m)}.$$

This can also be expanded to $6-3m$.

**Restriction:** The original denominator is zero when $m=2$, so the expression is defined only for $m\ne2$. Cancelling does not remove that restriction.

```python
import sympy as sp

m = sp.symbols("m")
original = 3 * (2 - m)**2 / (2 - m)
simplified = sp.cancel(original)

print(f"Simplified expression: {simplified}")
print("Factorised form: 3(2 - m), with m != 2")
assert sp.simplify(simplified - 3 * (2 - m)) == 0
```

### Python walkthrough for 7(a)

1. The symbol $m$ is created so SymPy can work with an algebraic expression.
2. The original fraction is built using powers and division.
3. cancel removes the common factor and prints $6-3m$, which equals $3(2-m)$.
4. The assertion checks the two forms are algebraically equal. The code also states $m\ne2$, because the original fraction cannot be evaluated at $m=2$.

### 7(b) Solve

Start with

$$7+x\leq\frac{5x}{2}-8.$$

Multiply every term by 2. Since 2 is positive, the inequality sign stays the same:

$$14+2x\leq5x-16.$$

Add 16 to both sides, then subtract $2x$:

$$30\leq3x.$$

Divide by 3, which is also positive:

$$\boxed{x\geq10}.$$

Check the boundary $x=10$: $7+10=17$ and $\frac{5(10)}{2}-8=17$, so equality is allowed.

```python
import sympy as sp

x = sp.symbols("x", real=True)
inequality = 7 + x <= 5 * x / 2 - 8
solution = sp.solve_univariate_inequality(inequality, x)

print(f"Solution: {solution}")
assert solution.as_set() == sp.Interval(10, sp.oo)
```

### Python walkthrough for 7(b)

1. The symbol $x$ is marked as real, then the inequality is entered exactly as written.
2. solve_univariate_inequality rearranges it and returns $x\geq10$.
3. The assertion checks that the result includes the boundary value 10.

### 7(c) Solve

Start with the compound inequality:

$$9<2y+4<12.$$

Subtract 4 from all three parts:

$$5<2y<8.$$

Divide all three parts by 2. Since 2 is positive, both inequality signs stay the same:

$$\boxed{\frac{5}{2}<y<4}.$$

Both inequalities are strict, so neither endpoint is included.

```python
import sympy as sp

y = sp.symbols("y", real=True)
lower_condition = 9 < 2 * y + 4
upper_condition = 2 * y + 4 < 12
solution = sp.reduce_inequalities([lower_condition, upper_condition], y)

print(f"Solution: {solution}")
assert solution.as_set() == sp.Interval.open(sp.Rational(5, 2), 4)
```

### Python walkthrough for 7(c)

## Question 8

$ABC$ is a right-angled triangle. $A$, $E$, $C$ are on a straight line, and $A$, $D$, $B$ are on a straight line. $ED$ is parallel to $CB$.

**(a)** Prove that triangle $ABC$ is similar to triangle $ADE$. (2 marks)

$ED=20$ cm, $CB=30$ cm and $AC=18$ cm.

**(b)** Work out the length of $EC$.

```python
import matplotlib.pyplot as plt

# Coordinates use the given lengths and preserve the triangle's proportions.
A = (0, 18)
C = (0, 0)
B = (30, 0)
E = (0, 6)
D = (20, 6)

fig, ax = plt.subplots(figsize=(9, 6))
def segment(p, q, width=2.1, color="#20242a"):
    ax.plot([p[0], q[0]], [p[1], q[1]],
            linewidth=width, color=color, solid_capstyle="round")

for p, q in [(A, B), (A, C), (C, B)]:
    segment(p, q)
segment(E, D, color="#376da5")

# Right-angle marker at C and matching parallel arrows.
ax.plot([0, 1, 1], [1, 1, 0], color="#20242a", linewidth=1.5)
for height in [0, 6]:
    ax.annotate("", xy=(17, height), xytext=(14, height),
                arrowprops=dict(arrowstyle="-|>", color="#376da5",
                                mutation_scale=16, linewidth=1.5))

for name, point, offset in [
    ("A", A, (-0.4, 0.7)), ("B", B, (0.3, -0.3)),
    ("C", C, (-0.9, -0.5)), ("D", D, (0.3, 0.0)),
    ("E", E, (-0.9, -0.2)),
]:
    ax.text(point[0] + offset[0], point[1] + offset[1],
            name, fontsize=16)

ax.annotate("", xy=(-2.5, 18), xytext=(-2.5, 0),
            arrowprops=dict(arrowstyle="<->", color="#4a4a4a"))
ax.text(-6.0, 9, "18 cm", fontsize=12, rotation=90, va="center")
ax.text(8, 6.5, "20 cm", fontsize=12)
ax.text(9, -1.3, "30 cm", fontsize=12)

ax.set_aspect("equal")
ax.set_xlim(-7, 33)
ax.set_ylim(-3, 21)
ax.axis("off")
ax.set_title("Question 8: similar triangles ABC and ADE", fontsize=14)
plt.show()
```

### 8(a) Prove the triangles are similar

The triangles share the angle at $A$: because $A$, $D$, $B$ are collinear and $A$, $E$, $C$ are collinear,

$$\angle BAC=\angle DAE.$$

Also, $ED\parallel CB$, so corresponding angles are equal:

$$\angle ABC=\angle ADE.$$

Two angles of triangle $ABC$ equal two angles of triangle $ADE$. Therefore, by **AA similarity**,

$$\boxed{\triangle ABC\sim\triangle ADE}.$$

The corresponding vertices are $A\leftrightarrow A$, $B\leftrightarrow D$ and $C\leftrightarrow E$.

### 8(b) Find $EC$

Corresponding sides of similar triangles have the same ratio. The smaller triangle has base $ED=20$ cm; the larger triangle has base $CB=30$ cm:

$$\frac{AE}{AC}=\frac{ED}{CB}=\frac{20}{30}=\frac{2}{3}.$$

As $AC=18$ cm,

$$AE=\frac{2}{3}\times18=12\text{ cm}.$$

$E$ is between $A$ and $C$, so

$$EC=AC-AE=18-12=\boxed{6\text{ cm}}.$$

Check: $\frac{AE}{AC}=\frac{12}{18}=\frac{2}{3}=\frac{20}{30}$.

```python
from fractions import Fraction

ED = 20
CB = 30
AC = 18

scale_factor = Fraction(ED, CB)
AE = scale_factor * AC
EC = AC - AE

print(f"Small-to-large scale factor = {scale_factor}")
print(f"AE = {AE} cm")
print(f"EC = {EC} cm")

assert AE == 12
assert EC == 6
assert Fraction(AE, AC) == Fraction(ED, CB)
```

### Python code walkthrough

1. The three given lengths are stored as $ED=20$, $CB=30$ and $AC=18$.
2. Fraction(ED, CB) calculates the small-to-large scale factor exactly: $20/30=2/3$.
3. Multiplying $AC$ by $2/3$ gives $AE=12$ cm.
4. Subtracting $AE$ from the whole side $AC$ gives $EC=6$ cm.
5. The assertions check both lengths and the equality of corresponding side ratios. The diagram code plots the triangles with matching parallel arrows and a right-angle marker.

## Question 10

Prove that the difference in the squares of two consecutive even numbers is always a multiple of 4. (3 marks)

### Worked proof

Let the two consecutive even numbers be $2k$ and $2k+2$, where $k$ is any integer. The larger square minus the smaller square is

$$
(2k+2)^2-(2k)^2.
$$

Expand and simplify:

$$
(2k+2)^2-(2k)^2
= (4k^2+8k+4)-4k^2
=8k+4
=\boxed{4(2k+1)}.
$$

Because $k$ is an integer, $2k+1$ is also an integer. The difference is therefore **4 times an integer**, so it is always a multiple of 4. This works for every pair of consecutive even numbers.

```python
import sympy as sp

k = sp.symbols("k", integer=True)
first_even = 2 * k
next_even = 2 * k + 2

difference = sp.expand(next_even**2 - first_even**2)
factored = sp.factor(difference)

print(f"Expanded difference: {difference}")
print(f"Factored difference: {factored}")

assert sp.simplify(difference - 4 * (2 * k + 1)) == 0
```

### Python code walkthrough

1. SymPy treats $k$ as an arbitrary integer, so the code checks the algebraic pattern rather than a few examples.
2. The two consecutive even numbers are written as $2k$ and $2k+2$.
3. expand works out the difference of their squares as $8k+4$.
4. factor rewrites this as $4(2k+1)$, making the factor of 4 visible.
5. The assertion confirms that the expanded expression equals $4(2k+1)$ for every $k$. Since $2k+1$ is an integer, this is a proof of divisibility by 4.

## Question 11

$T$ is inversely proportional to $w$.

$w$ is directly proportional to the cube root of $d$.

When $w=6$, $T=20$. When $w=1$, $d=8$.

Find the value of $d$ when $T=48$.

### Worked solution

Because $T$ is inversely proportional to $w$,

$$T=\frac{k}{w}.$$

Use $w=6$ and $T=20$ to find the constant:

$$20=\frac{k}{6}\quad\Rightarrow\quad k=120.$$

Hence $T=\frac{120}{w}$. When $T=48$,

$$48=\frac{120}{w}
\quad\Rightarrow\quad
w=\frac{120}{48}=\frac{5}{2}.$$

Now use the second proportionality. Because $w$ is directly proportional to $\sqrt[3]{d}$,

$$w=c\sqrt[3]{d}.$$

When $w=1$ and $d=8$, the cube root of $8$ is $2$, so

$$1=c\sqrt[3]{8}=2c
\quad\Rightarrow\quad c=\frac12.$$

Therefore $w=\frac12\sqrt[3]{d}$. Substitute $w=\frac52$:

$$\frac52=\frac12\sqrt[3]{d}
\quad\Rightarrow\quad
\sqrt[3]{d}=5
\quad\Rightarrow\quad
\boxed{d=125}.$$

Check: $d=125$ gives $w=\frac12(5)=2.5$, then $T=120\div2.5=48$.

```python
from fractions import Fraction

# Inverse proportionality: T = k / w.
k = 6 * 20
target_T = 48
target_w = Fraction(k, target_T)

# Direct proportionality: w = c * cube_root(d).
# When d = 8, cube_root(d) = 2.
c = Fraction(1, 2)
cube_root_d = target_w / c
d = cube_root_d**3

print(f"k = {k}")
print(f"w when T = 48: {target_w}")
print(f"cube root of d: {cube_root_d}")
print(f"d = {d}")

assert d == 125
assert Fraction(k, target_w) == target_T
assert c * cube_root_d == target_w
```

### Python code walkthrough

3. For direct proportionality, $w=c\sqrt[3]{d}$. The second data pair has $\sqrt[3]{8}=2$, so $c=1/2$.

| Time $t$ (s) | 0 | 2 | 4 | 6 |
|---|---:|---:|---:|---:|

Use a trapezium for each strip:

$$
\begin{aligned}
0\text{–}2\text{ s}:&\quad \tfrac12(0+30)\times2=30\text{ m},\\
2\text{–}4\text{ s}:&\quad \tfrac12(30+50)\times2=80\text{ m},\\
4\text{–}6\text{ s}:&\quad \tfrac12(50+60)\times2=110\text{ m}.

## Question 14

There are nine balls labelled 1 to 9 in a box. Lee takes two balls at random **without replacement**.

Lee says: “The probability that the sum of the numbers on the two balls will be an even number is greater than the probability that the product of the numbers will be an even number.”

Is Lee correct? Show how you get your answer.

### Worked solution

Among the numbers 1 to 9, there are **5 odd** balls ($1,3,5,7,9$) and **4 even** balls ($2,4,6,8$). The second ball is chosen from only 8 remaining balls.

A sum is even when both numbers are odd or both are even:

$$
\begin{aligned}
P(\text{both odd})&=\frac{5}{9}\times\frac{4}{8}=\frac{20}{72},\\
P(\text{both even})&=\frac{4}{9}\times\frac{3}{8}=\frac{12}{72}.
\end{aligned}
$$

These outcomes cannot happen together, so

$$
P(\text{even sum})
=\frac{20}{72}+\frac{12}{72}
=\frac{32}{72}
=\boxed{\frac49}.
$$

A product is even if **at least one** number is even. The only way for the product to be odd is to choose two odd numbers. Therefore,

$$
P(\text{even product})
=1-P(\text{both odd})
=1-\frac{20}{72}
=\frac{52}{72}
=\boxed{\frac{13}{18}}.
$$

Compare the two probabilities using denominator 18:

$$\frac49=\frac8{18}<\frac{13}{18}.$$

**Lee is not correct.** An even product is more likely than an even sum.

```python
from fractions import Fraction
from itertools import combinations

odd = 5
even = 4
total = 9

p_both_odd = Fraction(odd, total) * Fraction(odd - 1, total - 1)
p_both_even = Fraction(even, total) * Fraction(even - 1, total - 1)
p_even_sum = p_both_odd + p_both_even
p_even_product = 1 - p_both_odd

print(f"P(even sum) = {p_even_sum}")
print(f"P(even product) = {p_even_product}")
print(f"Is Lee correct? {p_even_sum > p_even_product}")

# Check by listing every unordered pair of different balls.
pairs = list(combinations(range(1, 10), 2))
even_sum_count = sum((a + b) % 2 == 0 for a, b in pairs)
even_product_count = sum((a * b) % 2 == 0 for a, b in pairs)
assert len(pairs) == 36
assert even_sum_count == 16
assert even_product_count == 26
assert p_even_sum == Fraction(even_sum_count, len(pairs)) == Fraction(4, 9)
assert p_even_product == Fraction(even_product_count, len(pairs)) == Fraction(13, 18)
```

### Python code walkthrough

1. The code stores the counts of odd, even and total balls. After one ball is taken, only 8 remain, which is why the second fractions use $total-1$.
2. Fraction keeps each probability exact. It calculates “both odd” and “both even”, then adds them for an even sum.
3. For an even product, it subtracts the chance of “both odd” from 1. This counts every pair with at least one even number.
4. The comparison prints False, so Lee is incorrect.
5. As a check, combinations lists the 36 unordered pairs of different balls. Exactly 16 have an even sum and 26 have an even product, matching the fractions $16/36=4/9$ and $26/36=13/18$.

The probability is an estimate because weights within each histogram class are treated as evenly spread.

```python
from fractions import Fraction

classes = [(0, 10, 3), (10, 30, 5), (30, 45, 8), (45, 60, 10)]
target_low, target_high = 20, 40

total_frequency = sum((right - left) * density
                      for left, right, density in classes)

target_frequency = 0
for left, right, density in classes:
    overlap_width = max(0, min(right, target_high) - max(left, target_low))
    target_frequency += overlap_width * density

probability = Fraction(target_frequency, total_frequency)
print(f"Total frequency = {total_frequency}")
print(f"Frequency from 20 g to 40 g = {target_frequency}")
print(f"Estimated probability = {probability} = {float(probability)}")

assert total_frequency == 400
assert target_frequency == 130
assert probability == Fraction(13, 40)
```

### Python code walkthrough

1. Each tuple in classes stores a bar's left boundary, right boundary and frequency density.
2. The total frequency is the sum of each bar's width multiplied by its density: 400.
3. For each bar, overlap_width finds how much of the class lies inside 20–40 g. Only the 10–30 g and 30–45 g bars overlap, each by 10 g.
4. Multiplying the overlap widths by their densities gives 50 and 80, so the target frequency is 130.
5. Fraction reduces $130/400$ exactly to $13/40$. The assertions check the totals and probability.

## Question 16

**(a)** Rationalise the denominator of $\dfrac{35}{\sqrt7}$. Give your answer in its simplest form. (2 marks)

**(b)** $\dfrac{\sqrt{27}-1}{2-\sqrt3}$ can be written in the form $a+b\sqrt3$, where $a$ and $b$ are integers. Work out $a$ and $b$. (4 marks)

### 16(a) Rationalise the denominator

Multiply the numerator and denominator by $\sqrt7$. This is multiplying by 1, so the value does not change:

$$
\frac{35}{\sqrt7}
=\frac{35\sqrt7}{\sqrt7\sqrt7}
=\frac{35\sqrt7}{7}
=\boxed{5\sqrt7}.
$$

The denominator is now a rational number, and the result is in simplest form.

### 16(b) Find $a$ and $b$

First simplify the numerator:

$$\sqrt{27}=\sqrt{9\times3}=3\sqrt3.$$

So the expression becomes

$$\frac{3\sqrt3-1}{2-\sqrt3}.$$

Multiply top and bottom by the **conjugate** $2+\sqrt3$:

$$
\frac{3\sqrt3-1}{2-\sqrt3}
=\frac{(3\sqrt3-1)(2+\sqrt3)}
       {(2-\sqrt3)(2+\sqrt3)}.
$$

The denominator is a difference of squares:

$$(2-\sqrt3)(2+\sqrt3)=2^2-(\sqrt3)^2=4-3=1.$$

Expand the numerator:

$$
(3\sqrt3-1)(2+\sqrt3)
=6\sqrt3+9-2-\sqrt3
=7+5\sqrt3.
$$

Therefore,

$$\boxed{a=7,\qquad b=5}.$$

Check: $(7+5\sqrt3)(2-\sqrt3)=3\sqrt3-1=\sqrt{27}-1$.

```python
import sympy as sp

root3 = sp.sqrt(3)
root7 = sp.sqrt(7)

part_a = 35 / root7
answer_a = sp.radsimp(part_a)
print(f"16(a): {answer_a}")

part_b = (sp.sqrt(27) - 1) / (2 - root3)
answer_b = sp.expand(sp.radsimp(part_b))
a = answer_b.coeff(root3, 0)
b = answer_b.coeff(root3)

print(f"16(b): {answer_b}")
print(f"a = {a}, b = {b}")

assert sp.simplify(answer_a - 5 * root7) == 0
assert sp.simplify(answer_b - (7 + 5 * root3)) == 0
assert (a, b) == (7, 5)
```

### Python code walkthrough

1. SymPy represents $\sqrt3$ and $\sqrt7$ exactly, without decimal rounding.
2. For 16(a), radsimp rationalises $35/\sqrt7$ and returns $5\sqrt7$.
3. For 16(b), the expression is entered as written. radsimp multiplies by a suitable conjugate; expand collects the ordinary and $\sqrt3$ terms.
4. coeff extracts the constant term $a=7$ and the coefficient of $\sqrt3$, $b=5$.
5. The assertions check both exact simplified expressions and the values of $a$ and $b$.

## Question 17

$g(x)=1-3x$ and $h(x)=2x^2-1$.

Show that $3gh(x)+hg(x)=0$ has just one solution for $x$. (5 marks)

Here $gh(x)$ means $g(h(x))$, and $hg(x)$ means $h(g(x))$.

### Worked solution

Start by finding both composite functions. For $gh(x)=g(h(x))$, put $h(x)=2x^2-1$ into $g$:

$$
g(h(x))
=1-3(2x^2-1)
=1-6x^2+3
=4-6x^2.
$$

For $hg(x)=h(g(x))$, put $g(x)=1-3x$ into $h$:

$$
\begin{aligned}
h(g(x))
&=2(1-3x)^2-1\\
&=2(1-6x+9x^2)-1\\
&=1-12x+18x^2.
\end{aligned}
$$

Substitute both into the given equation:

$$
3(4-6x^2)+(1-12x+18x^2)=0.
$$

Expand and collect terms:

$$
12-18x^2+1-12x+18x^2=0
\quad\Rightarrow\quad
13-12x=0.
$$

The $x^2$ terms cancel, leaving a **linear equation** with a non-zero coefficient of $x$. A linear equation of this form has exactly one solution:

$$
12x=13
\quad\Rightarrow\quad
\boxed{x=\frac{13}{12}}.
$$

Therefore, $3gh(x)+hg(x)=0$ has **just one solution**.

```python
import sympy as sp

x = sp.symbols("x")
def g(value):
    return 1 - 3 * value
def h(value):
    return 2 * value**2 - 1

gh = sp.expand(g(h(x)))
hg = sp.expand(h(g(x)))
equation_left = sp.expand(3 * gh + hg)
solutions = sp.solve(sp.Eq(equation_left, 0), x)

print(f"g(h(x)) = {gh}")
print(f"h(g(x)) = {hg}")
print(f"3g(h(x)) + h(g(x)) = {equation_left}")
print(f"Solutions: {solutions}")

assert gh == 4 - 6*x**2
assert hg == 1 - 12*x + 18*x**2
assert equation_left == 13 - 12*x
assert solutions == [sp.Rational(13, 12)]
```

### Python code walkthrough

1. The functions g and h accept a value or algebraic expression. Calling g(h(x)) and h(g(x)) builds the two compositions in the correct order.
2. expand multiplies out brackets, giving $4-6x^2$ and $1-12x+18x^2$.
3. Multiplying the first composition by 3 and adding the second cancels the $x^2$ terms, leaving $13-12x$.
4. solve finds $x=13/12$. The output list has one entry, matching the algebraic proof that this is a linear equation with one solution.
5. The assertions check each expansion and the exact solution.
