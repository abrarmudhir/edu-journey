## Question 1

### Find the **highest common factor (HCF)** of **54** and **120**.

The **Highest Common Factor (HCF)** is the largest number that divides exactly into both numbers.

#### Step 1: Factors of 54

$$
54 = 1 \times 54
$$

$$
54 = 2 \times 27
$$

$$
54 = 3 \times 18
$$

$$
54 = 6 \times 9
$$

So the factors of $54$ are:

$$
1,\ 2,\ 3,\ 6,\ 9,\ 18,\ 27,\ 54
$$

#### Step 2: Factors of 120

$$
120 = 1 \times 120
$$

$$
120 = 2 \times 60
$$

$$
120 = 3 \times 40
$$

$$
120 = 4 \times 30
$$

$$
120 = 5 \times 24
$$

$$
120 = 6 \times 20
$$

$$
120 = 8 \times 15
$$

$$
120 = 10 \times 12
$$

So the factors of $120$ are:

$$
1,\ 2,\ 3,\ 4,\ 5,\ 6,\ 8,\ 10,\ 12,\ 15,\ 20,\ 24,\ 30,\ 40,\ 60,\ 120
$$

#### Step 3: Find the common factors

The numbers that are factors of **both** 54 and 120 are:

$$
1,\ 2,\ 3,\ 6
$$

The **highest** of these numbers is:

$$
\boxed{6}
$$

Therefore:

$$
\boxed{\operatorname{HCF}(54,120)=6}
$$

```python
a = 54
b = 120

hcf = math.gcd(a, b)

print(f"The HCF of {a} and {b} is {hcf}.")
```

## Question 2

There are only **red counters, white counters, blue counters and green counters** in a bag.

Chris is going to take at random a counter from the bag.

The table shows the probability that he will take a red counter and the probability that he will take a white counter.

| Colour | Red | White | Blue | Green |
|---|---:|---:|---:|---:|
| Probability | 0.3 | 0.1 |  |  |

There are **twice as many blue counters as there are green counters** in the bag.

### (a)

Work out the probability that Chris will take a **blue counter**.

There are **45 red counters** in the bag.

### (a) Work out the probability of taking a blue counter

We know:

$$
P(\text{red}) = 0.3
$$

and

$$
P(\text{white}) = 0.1
$$

All probabilities must add up to:

$$
1
$$

### Step 1: Find the probability left for blue and green

First add the red and white probabilities:

$$
0.3 + 0.1 = 0.4
$$

Now subtract this from 1:

$$
1 - 0.4 = 0.6
$$

So the probability left for **blue and green together** is:

$$
P(\text{blue}) + P(\text{green}) = 0.6
$$

### Step 2: Use the ratio

There are **twice as many blue counters as green counters**.

So the ratio is:

$$
\text{Blue : Green} = 2 : 1
$$

There are:

$$
2 + 1 = 3
$$

equal parts altogether.

Divide \(0.6\) into 3 equal parts:

$$
0.6 \div 3 = 0.2
$$

So one part is:

$$
0.2
$$

### Step 3: Find the probability of blue

Blue represents **2 parts**, so:

$$
2 \times 0.2 = 0.4
$$

Therefore:

$$
\boxed{P(\text{blue}) = 0.4}
$$

### Answer

$$
\boxed{0.4}
$$

### (b)

Work out the **total number of counters** in the bag.

### (b) Work out the total number of counters in the bag

We are told there are **45 red counters**.

From the table:

$$
P(\text{red}) = 0.3
$$

Probability can be written as:

$$
P(\text{red}) =
\frac{\text{number of red counters}}
{\text{total number of counters}}
$$

Let the total number of counters be \(x\).

So:

$$
0.3 = \frac{45}{x}
$$

Multiply both sides by \(x\):

$$
0.3x = 45
$$

Now divide both sides by \(0.3\):

$$
x = \frac{45}{0.3}
$$

$$
x = 150
$$

Therefore, the total number of counters in the bag is:

$$
\boxed{150}
$$
